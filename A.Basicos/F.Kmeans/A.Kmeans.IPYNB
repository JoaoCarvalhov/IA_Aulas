No algoritmo de K-means, o número de clusters é definido pelo usuário, e o algoritmo tenta encontrar os centros dos clusters que minimizam a distância entre os pontos de dados e os centros dos clusters. 
O K-means é um algoritmo de aprendizado não supervisionado, pois não requer rótulos para os dados.


```python
from sklearn import datasets
from sklearn.metrics import confusion_matrix
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.decomposition import PCA

iris = datasets.load_iris()
print(iris.data)

kmeans = KMeans(n_clusters=3, n_init="auto")
labels = kmeans.fit_predict(iris.data)

print(labels)
print("------------------------")
print(confusion_matrix(iris.target, labels))

```

```python
# Visualização gráfica dos clusters no espaço 2D reduzido via PCA
pca = PCA(n_components=2)
iris_2d = pca.fit_transform(iris.data)

plt.figure(figsize=(14,6))
plt.subplot(1, 2, 1)
scatter = plt.scatter(iris_2d[:, 0], iris_2d[:, 1], c=labels, cmap='viridis', edgecolor='k', s=60)
plt.title('Clusters KMeans (3 clusters)')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.legend(*scatter.legend_elements(), title='Cluster')

plt.subplot(1, 2, 2)
scatter = plt.scatter(iris_2d[:, 0], iris_2d[:, 1], c=iris.target, cmap='Accent', edgecolor='k', s=60)
plt.title('Classes reais do Iris')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.legend(*scatter.legend_elements(), title='Classe')

plt.tight_layout()
plt.show()
```
