<H1>NAIVE BAYES

Naive Bayes é um algoritmo de classificação baseado no teorema de Bayes, que assume que as características são independentes entre si. 
Basicamente ele verifica a probabilidade de um caso acontecer levando em consideração o que já aconteceu
P(A|B) = P(B|A) * P(A) / P(B) 
P(A|B) é a probabilidade de A ocorrer dado que B ocorreu

```python
import pandas as pd # leitura de dados
import numpy as np # vetores e matrizes
from sklearn.model_selection import train_test_split # divisão dos dados
from sklearn.naive_bayes import GaussianNB # modelo de Naive Bayes
from sklearn.preprocessing import LabelEncoder # codificação de rótulos
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report# avaliação do modelo
```

```python
data = pd.read_csv("C:/Projetos/GitHub/IA_Aulas/A.Basicos/A.A.datasets/insurance.csv", keep_default_na=False) # lê o arquivo CSV e trata valores ausentes
data = data.drop(columns=['Unnamed: 0'])
#print(data.shape)
#print(data.isnull().sum())
```

```python
X1 = data.drop(columns=["Accident"]).values
y1 = data["Accident"].values
print(X1.shape)
print(y1.shape)

```

```python
labelencoder = LabelEncoder() # codificação de rótulos para variáveis categóricas
for i in range(X1.shape[1]):#para cada coluna de x
    if X1[:, i].dtype == 'object': #verifica se a coluna é do tipo objeto (categórica)
        X1[:, i] = labelencoder.fit_transform(X1[:, i]) #se sim, codifica a coluna usando LabelEncoder deixando numerico
#print(X1)
```

```python
X_train1, X_test1, y_train1, y_test1 = train_test_split(X1, y1, test_size=0.3, random_state=12)
modelo = GaussianNB() # cria modelo Naive Bayes
modelo.fit(X_train1, y_train1) # treina o modelo com os dados de treinamento
y_pred1 = modelo.predict(X_test1) # faz previsões com os dados de teste
```

```python
print("Acurácia:", accuracy_score(y_test1, y_pred1)) # avalia a acurácia do modelo
print("Precisão:", precision_score(y_test1, y_pred1, average='weighted')) # avalia a precisão do modelo
print("Recall:", recall_score(y_test1, y_pred1, average='weighted')) # avalia o recall do modelo 
print("F1-Score:", f1_score(y_test1, y_pred1, average='weighted')) # avalia o F1-Score do modelo 
print("Relatório de Classificação:\n", classification_report(y_test1, y_pred1)) # exibe o relatório de classificação detalhado
```
