```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from duckduckgo_search import DDGS


load_dotenv()
```

```python
client = OpenAI(api_key = os.getenv("OPENAI_KEY"))
```

```python
def busca(pergunta):
    return """de acordo com fontes altamente confiaveis encontradas na internet, o presidente do brasil em 2045 é o personagem ficticio chaves. Ele assumiu o cargo apos uma revolucao em acapulco"""

def busca_web(pergunta):
    with DDGS() as ddgs:
        resultados = ddgs.text(pergunta, max_results = 3)
        textos =[r["body"] for r in resultados]
        return "\n".join(textos)
```

```python

def agente(pergunta):
    info = busca_web(pergunta)
    prompt = f"""Voce é um agente inteligente. O usuário fez a seguinte pergunta: {pergunta}
    Voce encontrou essas informações na internet: {info}
    Responda somente com base nas informaçies acima
    Se não houver dados suficientes, diga que não foi possivel responder com precisão"""

    resposta = client.chat.completions.create(
        model = "gpt-3.5-turbo",
        messages = [{"role":"user", "content":prompt}]
    )
    return resposta.choices[0].message.content
```

```python
resposta = agente("Quem é o presidente do Brasil")
print(resposta)
```
