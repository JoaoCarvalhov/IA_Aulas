```python
import cv2
import numpy as np
```

```python
Model_config = "/home/arkade01/projetos/IA_Aulas/C.RNA/datasets/MobileNetSSD_deploy.prototxt"
Model_Weights = "/home/arkade01/projetos/IA_Aulas/C.RNA/datasets/MobileNetSSD_deploy.caffemodel"
Classes = [
    "background", "aeroplane", "bicycle", "bird", "boat",
    "bottle", "bus", "car", "cat", "chair", "cow",
    "diningtable", "dog", "horse", "motorbike",
    "person", "pottedplant", "sheep", "sofa",
    "train", "tvmonitor"
]
```

```python
net = cv2.dnn.readNetFromCaffe(Model_config,Model_Weights)
```

```python
def detect_person_in_video(video_path):
    cap = cv2.VideoCapture(video_path)
    frame_count = 0
    while True:
        ret,frame = cap.read()
        if not ret:
            print("Cabou o video :(")
            break
        frame_count +=1
        (h,w) = frame.shape[:2]
        blob = cv2.dnn.blobFromImage(frame, 0.007843,(300,300), 127.5)  #a função blobFromImage é usada para pré-processar a imagem antes de alimentá-la na rede neural. Ela realiza as seguintes operações:
                                                                        #Redimensionamento: A imagem é redimensionada para um tamanho específico (neste caso, 300x300 pixels) para garantir que a entrada da rede neural seja consistente.
                                                                        #Subtração da média: A função subtrai um valor médio (neste caso, 127.5) de cada pixel da imagem para normalizar os valores dos pixels.
                                                                        #Escala: A função escala os valores dos pixels para um intervalo específico (neste caso, 0.007843 ou 1 / 127.5) para garantir que os valores dos pixels estejam em um intervalo adequado

        net.setInput(blob) # aqui tem a preparação da imagem para a rede neural
        detection = net.forward()   #  A rede neural retorna um tensor no formato: (1, 1, N, 7), onde # 1 -> batch, 1 -> dimensão auxiliar do modelo, N -> quantidade de detecções, 7 -> atributos de cada detecção 
                                    #Cada detecção possui os atributos: [batch_id, class_id, confidence, x1, y1, x2, y2]

        for i in range(detection.shape[2]): # detection: (1, 1, Numero de detecções, 7) -> detection.shape[2] é o número de detecções
            confidence = detection[0,0,i,2]
            if confidence > 0.9:
                idx = int(detection[0,0,i,1])
                if Classes[idx] == "person":
                    box = detection[0,0,i,3:7] * np.array([w,h,w,h])
                    (startX, startY, endX, endY) = box.astype("int")

                    label = f"{Classes[idx]}: {confidence*100:.2f}%"
                    cv2.rectangle(frame, (startX,startY), (endX, endY), (0,255,0),2)
                    cv2.putText(frame, label, (startX,startY-10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,0), 2)
                    #cv2.imshow("Detecção", frame)
                    
                    detected_frame_path = f"detected_frame_{frame_count}.jpg"
                    cv2.imwrite(detected_frame_path, frame)

                    cap.release()
                    return "pessoa detectada no video"
    cap.release()
    return "pessoa não detectada no video"
                
```

```python
print(detect_person_in_video("/home/arkade01/projetos/IA_Aulas/C.RNA/datasets/people.mp4"))
```
