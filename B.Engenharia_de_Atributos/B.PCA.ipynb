```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn import datasets
```

```python
iris = datasets.load_iris()
previsores = iris.data
classe = iris.target
previsores
```

```python
ss = StandardScaler()
previsores = ss.fit_transform(previsores)

previsores
```

```python
X_train, X_test, y_train, y_test = train_test_split(previsores, classe, random_state=123, test_size=0.3)

floresta = RandomForestClassifier(n_estimators = 100, random_state= 1234)
floresta.fit(X_train, y_train)
previsoes = floresta.predict(X_test)
accuracy_score(y_test, previsoes)
```

<h2> PCA
> Testando com PCA antes do split

```python
pca = PCA(n_components=3)
previsores2 = pca.fit_transform(iris.data)
previsores2
```

```python
X_train2, X_test2, y_train2, y_test2 = train_test_split(previsores2, classe, test_size=0.3, random_state=123)

floresta2 = RandomForestClassifier()

floresta2.fit(X_train2, y_train2)

previsoes2 = floresta2.predict(X_test2)
```

```python
print(accuracy_score(previsoes2, y_test2))
```

<h2>Fazendo agora com o pca após o split

```python
previsores3 = iris.data
classe3 = iris.target

pca3 = PCA()
X_train3, X_test3, y_train3, y_test3 = train_test_split(previsores3, classe3, random_state=123, test_size=0.3)

X_train3_pca = pca3.fit_transform(X_train3)
X_test3_pca = pca3.transform(X_test3)
```

```python
rf3 = RandomForestClassifier(n_estimators=100, random_state=1234)
rf3.fit(X_train3_pca, y_train3)
previsoes3 = rf3.predict(X_test3_pca)
```

```python
print(accuracy_score(previsoes3, y_test3))
```
