<h1> Classes desbalanceadas


Exemplos como transações fraudulentas ou celulas infectadas normamlmente não são a maioria dos dados que se guarda 

Então seus datasets seão desbalanceados. Então quais são as soluções?

Undersampling (diminuir classe majoritaria) ou oversampling (Aumenta4r classe minoritaria [com smote, por exemplo])

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from imblearn.over_sampling import SMOTENC
```

```python
credito = pd.read_csv("C:\\Projetos\\GitHub\\IA_Aulas\\B.Engenharia_de_Atributos\\datasets\\credit_simple.csv", sep=";")
credito
```

```python
credito.groupby(["CLASSE"]).size()
```

```python
X = credito.drop(columns=["CLASSE"])
y = credito["CLASSE"]

#tirando valores nulos
mediana = X["SALDO_ATUAL"].median()
X["SALDO_ATUAL"] = X["SALDO_ATUAL"].fillna(mediana)
X["ESTADOCIVIL"] = X["ESTADOCIVIL"].fillna("masculino solteiro")

X = X.values
y = y.values
```

```python
for i in range(X.shape[1]):
    ll = LabelEncoder()
    # Verifica se primeiro elemento é string (não só dtype)
    if isinstance(X[0, i], str):
        X[:, i] = ll.fit_transform(X[:, i]) 
```

```python
pd.DataFrame(X)
```

```python
y
```

```python
sm = SMOTENC(random_state=0, categorical_features=[3,5,6])
X_res, y_res = sm.fit_resample(X, y)
```
